Part 2 — imports, logging, config loader, constants

```python
# UPDATED
import os, json, logging, joblib, yaml
import numpy as np
import pandas as pd
from datetime import datetime

from sklearn.model_selection import train_test_split, RepeatedStratifiedKFold
from sklearn.metrics import f1_score, classification_report, confusion_matrix
from sklearn.utils.multiclass import unique_labels
from sklearn.base import BaseEstimator, TransformerMixin, ClassifierMixin
from sklearn.preprocessing import PowerTransformer
from sklearn.pipeline import Pipeline as SkPipeline
from sklearn.preprocessing import FunctionTransformer

from imblearn.pipeline import Pipeline as ImbPipeline  # UPDATED
from imblearn.combine import SMOTETomek

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier

import matplotlib.pyplot as plt
import seaborn as sns

# Logging setup (replace prints)  # UPDATED
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)
log = logging.getLogger("eeg")

# Load config  # UPDATED
with open("config.yaml", "r") as f:
    CFG = yaml.safe_load(f)

DATA_PATH = CFG["paths"]["data_path"]
OUT_DIR   = CFG["paths"]["out_dir"]
PIPE_PATH = CFG["paths"]["pipeline_path"]
RANDOM_STATE = int(CFG["random_state"])
FEATS_USED = CFG["features_used"]
CV_SPLITS = int(CFG["cv"]["n_splits"])
CV_REPEATS = int(CFG["cv"]["n_repeats"])
EARLY_CFG = CFG.get("early_stopping", {"enabled": True, "rounds": 50})

os.makedirs(OUT_DIR, exist_ok=True)
```

Part 3 — helpers: checks, engineered features, rare-class utils

```python
# UPDATED
def check_columns(df: pd.DataFrame, required: list):
    missing = [c for c in required if c not in df.columns]
    if missing:
        have = set(df.columns.tolist())
        need = set(required)
        diff = {
            "missing": sorted(list(need - have)),
            "unexpected": sorted(list(have - need))
        }
        msg = f"Required EEG columns missing. Diff: {diff}"
        log.error(msg)
        raise AssertionError(msg)

# Engineered features built *after* PowerTransformer
def engineer_from_Xt(Xt: np.ndarray):
    delta, theta, a_low, a_high, b_low, b_high, g_low, g_high = [Xt[:, i] for i in range(8)]
    alpha_mean = (a_low + a_high) / 2.0
    beta_mean  = (b_low + b_high) / 2.0
    gamma_mean = (g_low + g_high) / 2.0
    slow_sum  = delta + theta + alpha_mean
    fast_sum  = beta_mean + gamma_mean
    total_pow = slow_sum + fast_sum

    eps = 1e-9
    beta_gamma_ratio   = beta_mean / (gamma_mean + eps)
    fast_to_slow_ratio = fast_sum  / (slow_sum  + eps)
    beta_to_alpha      = beta_mean / (alpha_mean + eps)
    gamma_to_alpha     = gamma_mean/ (alpha_mean + eps)
    theta_to_delta     = theta     / (delta      + eps)

    fast_prod      = beta_mean * gamma_mean
    alpha_slow_mix = alpha_mean * (theta + delta)

    extras = np.column_stack([
        alpha_mean, beta_mean, gamma_mean, slow_sum, fast_sum, total_pow,
        beta_gamma_ratio, fast_to_slow_ratio, beta_to_alpha, gamma_to_alpha,
        theta_to_delta, fast_prod, alpha_slow_mix
    ])
    extra_names = [
        "alpha_mean","beta_mean","gamma_mean","slow_sum","fast_sum","total_pow",
        "beta_gamma_ratio","fast_to_slow_ratio","beta_to_alpha","gamma_to_alpha",
        "theta_to_delta","fast_prod","alpha_slow_mix"
    ]
    Xeng = np.column_stack([Xt, extras])
    return Xeng, extra_names

class FeatureEngineer(BaseEstimator, TransformerMixin):  # UPDATED
    """Fits PowerTransformer on train-only; transforms + engineers features."""
    def __init__(self, method='yeo-johnson'):
        self.method = method
        self.pt_ = None
        self.feature_names_ = None

    def fit(self, X, y=None):
        self.pt_ = PowerTransformer(method=self.method)
        Xt = self.pt_.fit_transform(X)
        _, extra_names = engineer_from_Xt(Xt)
        base_names = [f"pt_{name}" for name in FEATS_USED]
        self.feature_names_ = base_names + extra_names
        return self

    def transform(self, X):
        Xt = self.pt_.transform(X)
        Xeng, _ = engineer_from_Xt(Xt)
        return Xeng

    def get_feature_names_out(self):
        return np.array(self.feature_names_)

def detect_rare_class(y: np.ndarray) -> int:  # UPDATED
    """Auto-detect the rarest label among present classes."""
    vals, counts = np.unique(y, return_counts=True)
    rare_idx = np.argmin(counts)
    return int(vals[rare_idx])

def make_targeted_weights(y: np.ndarray, weight_map: dict) -> np.ndarray:  # UPDATED
    """Map of class->multiplier; classes not present ignored."""
    w = np.ones_like(y, dtype=float)
    for cls, m in weight_map.items():
        cls = int(cls)
        w[y == cls] = float(m)
    return w
```

Part 4 — ensemble wrapper (soft vote) that works in a Pipeline + early stopping

```python
# UPDATED
class EnsembleSoftVoter(ClassifierMixin, BaseEstimator):
    """
    sklearn-compatible wrapper that trains XGB + LGBM and soft-votes their probabilities.
    Accepts sample_weight via fit(..., sample_weight=...), and early stopping params via
    fit(..., eval_set=(X_val, y_val)) when available (no leakage: used only in train/val).
    """
    def __init__(self, n_classes, xgb_params: dict, lgbm_params: dict, random_state=42):
        self.n_classes = int(n_classes)
        self.xgb_params = dict(xgb_params)
        self.lgbm_params = dict(lgbm_params)
        self.random_state = int(random_state)
        self.xgb_ = None
        self.lgbm_ = None
        self.classes_ = None

    def fit(self, X, y, sample_weight=None, eval_set=None, early_stopping_rounds=None, verbose=False):
        # Build models fresh each fit
        xgb_kw = dict(self.xgb_params)
        lgb_kw = dict(self.lgbm_params)

        # Ensure multiclass setup consistent with original behavior
        xgb_kw.update(dict(objective='multi:softprob', num_class=self.n_classes, random_state=self.random_state))
        lgb_kw.update(dict(objective='multiclass', num_class=self.n_classes, random_state=self.random_state))

        self.xgb_  = XGBClassifier(**xgb_kw)
        self.lgbm_ = LGBMClassifier(**lgb_kw)

        # Try early stopping if requested and eval_set provided; silently fallback
        if eval_set is not None and early_stopping_rounds is not None:
            try:
                self.xgb_.fit(
                    X, y,
                    sample_weight=sample_weight,
                    eval_set=[eval_set],
                    early_stopping_rounds=early_stopping_rounds,
                    verbose=verbose
                )
            except TypeError:
                self.xgb_.fit(X, y, sample_weight=sample_weight)
            try:
                self.lgbm_.fit(
                    X, y,
                    sample_weight=sample_weight,
                    eval_set=[eval_set],
                    callbacks=[],
                    verbose=verbose
                )
            except TypeError:
                self.lgbm_.fit(X, y, sample_weight=sample_weight)
        else:
            self.xgb_.fit(X, y, sample_weight=sample_weight)
            self.lgbm_.fit(X, y, sample_weight=sample_weight)

        self.classes_ = np.unique(y)
        return self

    def predict_proba(self, X):
        px = self.xgb_.predict_proba(X)
        pl = self.lgbm_.predict_proba(X)
        if px.shape != pl.shape:
            raise RuntimeError("XGB and LGBM prob shapes mismatch")
        return (px + pl) / 2.0

    def predict(self, X):
        p = self.predict_proba(X)
        return p.argmax(axis=1)
```

Part 5 — load data, validate columns, split holdout


```python
# UPDATED
if not os.path.exists(DATA_PATH):
    raise FileNotFoundError(f"File not found: {DATA_PATH}")

df = pd.read_csv(DATA_PATH)
log.info(f"Loaded data shape: {df.shape}")
check_columns(df, FEATS_USED + ["class"])  # robust diff  # UPDATED

X_all = df[FEATS_USED].values
y_all = df["class"].values
n_classes = len(np.unique(y_all))
if n_classes < 3:
    log.warning("n_classes < 3 detected; some tuning features will be skipped safely.")  # UPDATED
log.info(f"Classes found: {np.unique(y_all)}")

# Holdout split (80/20) with stratification (same as original)
X_tr_full, X_te, y_tr_full, y_te = train_test_split(
    X_all, y_all, test_size=0.2, stratify=y_all, random_state=RANDOM_STATE
)
log.info(f"Train/TEST shapes: {X_tr_full.shape} / {X_te.shape}")
```

Part 6 — leakage-proof CV (5×3) using FE + SMOTETomek + ensemble

```python
# UPDATED: in your CV loop, replace the ImbPipeline with the wrapper estimator
fe = FeatureEngineer(method='yeo-johnson')
sampler = SMOTETomek(random_state=RANDOM_STATE)

xgb_params = dict(CFG["models"]["xgb"])
lgbm_params = dict(CFG["models"]["lgbm"]) # Added lgbm_params

# Initialize lists to store CV scores
cv_scores = []

cv = RepeatedStratifiedKFold(n_splits=CV_SPLITS, n_repeats=CV_REPEATS, random_state=RANDOM_STATE) # Defined cv

for fold_idx, (idx_tr, idx_va) in enumerate(cv.split(X_tr_full, y_tr_full)): # Added fold_idx
    log.info(f"Processing fold {fold_idx+1}/{CV_SPLITS*CV_REPEATS}") # Added log

    X_raw_tr, X_raw_va = X_tr_full[idx_tr], X_tr_full[idx_va]
    y_tr,       y_va   = y_tr_full[idx_tr], y_tr_full[idx_va]

    # Fit FE on the current training fold
    fe_fold = FeatureEngineer(method='yeo-johnson').fit(X_raw_tr)
    X_tr_fe = fe_fold.transform(X_raw_tr)
    X_va_fe = fe_fold.transform(X_raw_va)

    # Resample the feature-engineered data
    X_tr_bal, y_tr_bal = sampler.fit_resample(X_tr_fe, y_tr)

    # auto-detect rare class per fold (optional)
    rare_cls = detect_rare_class(y_tr_bal)
    # Use the maximum rare weight candidate found in the config for this rare class
    rare_weight_candidates = CFG.get("rare_weight_candidates", [1.0, 2.0, 3.0, 4.0])
    weight_map = {rare_cls: max(rare_weight_candidates)}

    # Create sample weights based on the resampled data and the weight map
    w_tr = make_targeted_weights(y_tr_bal, weight_map)

    # Use the EnsembleSoftVoter directly
    est = EnsembleSoftVoter(
        n_classes=n_classes,
        xgb_params=xgb_params,
        lgbm_params=lgbm_params,
        random_state=RANDOM_STATE
    )

    # Fit the ensemble on the resampled and weighted data
    # No early stopping during CV unless explicitly configured and eval_set is provided
    est.fit(X_tr_bal, y_tr_bal, sample_weight=w_tr)

    y_va_pred = est.predict(X_va_fe) # Predict on the feature-engineered validation data
    fold_f1 = f1_score(y_va, y_va_pred, average="macro", zero_division=0)
    cv_scores.append(fold_f1)
    log.info(f"Fold {fold_idx+1} Macro-F1: {fold_f1:.4f}") # Added log

# Log the mean and std of CV scores
log.info(f"CV Mean Macro-F1: {np.mean(cv_scores):.4f} +/- {np.std(cv_scores):.4f}")
```

Part 7 — final FE + train/val tuning (safe multipliers + optional early stopping), then test

```python
# UPDATED
# Fit FE on full TRAIN (still leakage-proof)
fe_full = FeatureEngineer().fit(X_tr_full)
X_trfe_full = fe_full.transform(X_tr_full)
X_te_fe     = fe_full.transform(X_te)

# Final train/val split for optional tuning (no test leakage)
X_tr_df, X_val_df, y_tr, y_val = train_test_split(
    pd.DataFrame(X_trfe_full, columns=fe_full.get_feature_names_out()),
    y_tr_full,
    test_size=0.2, stratify=y_tr_full, random_state=RANDOM_STATE
)
log.info(f"Final train/val shapes: {X_tr_df.shape} / {X_val_df.shape}")

# Build sampler + classifier (we’ll fit on engineered features)
SMOTETomek(random_state=RANDOM_STATE, sampling_strategy="not minority")
X_tr_bal, y_tr_bal = sampler.fit_resample(X_tr_df.values, y_tr)
log.info(f"Resampled train size: {X_tr_bal.shape[0]}")

# Auto rare-class detection on the resampled labels’ ORIGINAL mapping (use y_tr_bal)  # UPDATED
rare_cls = detect_rare_class(y_tr_bal)
log.info(f"Auto-detected rare class: {rare_cls}")

def eval_with_weights(weight_map: dict):
    """Train both models with given class->multiplier weight map and return val macro-F1."""
    w_tr = make_targeted_weights(y_tr_bal, weight_map)

    clf = EnsembleSoftVoter(n_classes=n_classes, xgb_params=xgb_params, lgbm_params=lgbm_params, random_state=RANDOM_STATE)

    # Optional early stopping on (X_val_df, y_val) — safe, no leakage into test
    use_es = EARLY_CFG.get("enabled", True)
    es_rounds = int(EARLY_CFG.get("rounds", 50))
    eval_set = (X_val_df.values, y_val)

    if use_es:
        clf.fit(X_tr_bal, y_tr_bal, sample_weight=w_tr, eval_set=eval_set, early_stopping_rounds=es_rounds, verbose=False)
    else:
        clf.fit(X_tr_bal, y_tr_bal, sample_weight=w_tr)

    y_val_pred = clf.predict(X_val_df.values)
    score = f1_score(y_val, y_val_pred, average="macro", zero_division=0)
    return score, clf

best_score, best_clf, best_map = -1.0, None, {}

# 1) Simple rare-weight candidates (for the rare class only)
if CFG.get("enable_multiplier_tuning", True):  # UPDATED
    for rw in CFG.get("rare_weight_candidates", [1.0, 2.0, 3.0, 4.0]):
        wmap = {rare_cls: float(rw)}
        score, clf = eval_with_weights(wmap)
        log.info(f"Tuning (rare={rare_cls}, weight={rw}) -> val Macro-F1={score:.4f}")
        if score > best_score:
            best_score, best_clf, best_map = score, clf, wmap

# 2) Optional full multiplier grids (safe indices only)
grids = CFG.get("multiplier_grids", [])
if n_classes < 3 and grids:
    log.info("n_classes < 3; skipping multiplier grid tuning safely.")  # UPDATED
elif grids:
    for grid in grids:
        # keep only classes present in y_tr_bal
        present = set(np.unique(y_tr_bal).tolist())
        safe_map = {int(k): float(v) for k, v in grid.items() if int(k) in present}
        if not safe_map:
            continue
        score, clf = eval_with_weights(safe_map)
        log.info(f"Tuning grid {safe_map} -> val Macro-F1={score:.4f}")
        if score > best_score:
            best_score, best_clf, best_map = score, clf, safe_map

# Fallback if no tuning happened
if best_clf is None:
    log.info("No tuning applied; using neutral weights.")
    best_map = {}
    _, best_clf = eval_with_weights(best_map)

log.info(f"Selected weight map: {best_map} with val Macro-F1={best_score:.4f}")

# Test inference (FE was fit on full train earlier; safe)
y_te_pred = best_clf.predict(X_te_fe)
known = set(unique_labels(y_all))
pred_set = set(np.unique(y_te_pred))
if not pred_set.issubset(known):
    raise AssertionError(f"Predictions contain unknown labels: {pred_set - known}")

macro_f1 = f1_score(y_te, y_te_pred, average='macro', zero_division=0)
cm = confusion_matrix(y_te, y_te_pred, labels=np.sort(np.array(list(known))))
report = classification_report(y_te, y_te_pred, zero_division=0)
log.info(f"=== Test Set === Macro-F1: {macro_f1:.4f}")
```

Part 9 — persist a full inference pipeline and add a tiny predict.py

```python
# UPDATED
# Build final inference pipeline object (FE + trained ensemble)
class InferencePipeline(BaseEstimator):  # thin wrapper to bundle FE and clf
    def __init__(self, fe, clf, feature_names):
        self.fe = fe
        self.clf = clf
        self.feature_names = feature_names

    def predict(self, X):           # X = raw array with FEATS_USED order
        X_fe = self.fe.transform(X)
        return self.clf.predict(X_fe)

    def predict_proba(self, X):
        X_fe = self.fe.transform(X)
        return self.clf.predict_proba(X_fe)

full_pipe = InferencePipeline(fe_full, best_clf, FEATS_USED)
joblib.dump(full_pipe, PIPE_PATH)
log.info(f"Saved pipeline to {PIPE_PATH}")

# Write a tiny predict.py helper  # UPDATED
predict_py = f'''
import sys, joblib, pandas as pd
from sklearn.utils.multiclass import unique_labels

PIPE_PATH = "{PIPE_PATH}"
FEATS_USED = {FEATS_USED}

def main(csv_path):
    pipe = joblib.load(PIPE_PATH)
    df = pd.read_csv(csv_path)
    missing = [c for c in FEATS_USED if c not in df.columns]
    if missing:
        raise AssertionError(f"Missing required columns: {{missing}}")
    X = df[FEATS_USED].values
    preds = pipe.predict(X)
    out = df.copy()
    out["pred"] = preds
    out.to_csv("predictions.csv", index=False)
    print("Wrote predictions.csv")

if __name__ == "__main__":
    if len(sys.argv) != 2:
        print("Usage: python predict.py <path_to_csv>")
        sys.exit(1)
    main(sys.argv[1])
'''
with open("predict.py", "w") as f:
    f.write(predict_py)
print("Wrote predict.py")
```
