```python
# eeg_pipeline.py  — dataset-aware version
from __future__ import annotations
import os, json
from dataclasses import dataclass, asdict
from typing import Tuple, List, Dict, Any
from datetime import datetime

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    accuracy_score, f1_score, roc_auc_score,
    confusion_matrix, classification_report
)
from imblearn.combine import SMOTETomek
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.calibration import CalibratedClassifierCV
import joblib
import matplotlib.pyplot as plt
```

```python
# ---------------- Part 1: Configuration ----------------
@dataclass
class Config:
    data_path: str = "/content/modal_train_data_avg(1).csv"
    target_col: str = "class"
    test_size: float = 0.2
    random_state: int = 42
    scale: bool = True
    smote: bool = True
    model_dir: str = "./artifacts"
    author: str = "user"
    # model toggles
    calibrate: bool = True
```

```python
# ---------------- Part 2: Data Loading & Validation ----------------
EXPECTED_COLUMNS = [
    "GOOD SIGNAL(0-100)", "ATTENTION(0-100)", "MEDITATION(0-100)",
    "DELTA", "THETA", "LOW ALPHA", "HIGH ALPHA",
    "LOW BETA", "HIGH BETA", "LOW GAMMA", "HIGH GAMMA", "class"
]

def load_data(cfg: Config) -> pd.DataFrame:
    df = pd.read_csv(cfg.data_path)
    return df

def validate(df: pd.DataFrame, target_col: str) -> None:
    assert target_col in df.columns, f"Missing target column: {target_col}"
    # Check at least some of the expected EEG bands exist
    must_have = {"DELTA","THETA","LOW ALPHA","HIGH ALPHA","LOW BETA","HIGH BETA","LOW GAMMA","HIGH GAMMA"}
    assert must_have.issubset(set(df.columns)), "One or more EEG band columns missing."
    # Impute numeric NAs with median
    if df.isna().any().any():
        df.fillna(df.median(numeric_only=True), inplace=True)
```

```python
# ---------------- Part 3: Feature Engineering ----------------
BANDS = {
    "alpha_low": "LOW ALPHA", "alpha_high": "HIGH ALPHA",
    "beta_low": "LOW BETA", "beta_high": "HIGH BETA",
    "gamma_low": "LOW GAMMA", "gamma_high": "HIGH GAMMA",
    "theta": "THETA", "delta": "DELTA"
}
NON_SIGNAL_FEATS = ["ATTENTION(0-100)", "MEDITATION(0-100)"]
SIGNAL_QUALITY = "GOOD SIGNAL(0-100)"
EPS = 1e-6

class FeatureEngineer:
    def __init__(self):
        self.feature_names_: List[str] | None = None

    def transform(self, df: pd.DataFrame, target_col: str) -> pd.DataFrame:
        X = df.drop(columns=[target_col]).copy()
        y = df[target_col].copy()

        # Aggregated band powers
        alpha_avg = (X[BANDS["alpha_low"]] + X[BANDS["alpha_high"]]) / 2.0
        beta_avg  = (X[BANDS["beta_low"]]  + X[BANDS["beta_high"]])  / 2.0
        gamma_avg = (X[BANDS["gamma_low"]] + X[BANDS["gamma_high"]]) / 2.0

        # Ratios commonly used in EEG literature
        X["ratio_alpha_beta"]  = alpha_avg / (beta_avg + EPS)
        X["ratio_theta_beta"]  = X[BANDS["theta"]] / (beta_avg + EPS)
        X["ratio_gamma_beta"]  = gamma_avg / (beta_avg + EPS)
        X["ratio_delta_theta"] = X[BANDS["delta"]] / (X[BANDS["theta"]] + EPS)

        # Normalize attention/meditation by signal quality to damp artifacts
        X["attn_norm"]  = X["ATTENTION(0-100)"]   / (X[SIGNAL_QUALITY] + EPS)
        X["med_norm"]   = X["MEDITATION(0-100)"]  / (X[SIGNAL_QUALITY] + EPS)

        # Row-wise stats across the eight raw band channels
        raw_bands = [
            BANDS["delta"], BANDS["theta"],
            BANDS["alpha_low"], BANDS["alpha_high"],
            BANDS["beta_low"],  BANDS["beta_high"],
            BANDS["gamma_low"], BANDS["gamma_high"]
        ]
        X["band_mean"] = X[raw_bands].mean(axis=1)
        X["band_std"]  = X[raw_bands].std(axis=1).fillna(0.0)

        out = pd.concat([X, y], axis=1)
        self.feature_names_ = [c for c in out.columns if c != target_col]
        return out
```

```python
# ---------------- Part 4: Split & Class Weights ----------------
def split_and_weights(
    df: pd.DataFrame, target_col: str, test_size: float, random_state: int
):
    X = df.drop(columns=[target_col]).values
    y = df[target_col].values

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )

    classes = np.unique(y_train)
    weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_train)
    class_weight = {int(c): float(w) for c, w in zip(classes, weights)}

    # For XGBoost binary: scale_pos_weight = w_pos / w_neg
    scale_pos_weight = None
    if len(classes) == 2:
        w_neg = class_weight[int(classes[0])]
        w_pos = class_weight[int(classes[1])]
        scale_pos_weight = w_pos / (w_neg + 1e-12)

    return X_train, X_test, y_train, y_test, class_weight, classes, scale_pos_weight
```

```python
# ---------------- Part 5: Preprocessing ----------------
class Preprocessor:
    def __init__(self, scale: bool = True, smote: bool = True, random_state: int = 42):
        self.scale = scale
        self.smote = smote
        self.scaler = StandardScaler() if scale else None
        self.smotetomek = SMOTETomek(random_state=random_state) if smote else None

    def fit_transform(self, X: np.ndarray, y: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
        Xp, yp = X, y
        if self.scaler:
            Xp = self.scaler.fit_transform(Xp)
        if self.smotetomek:
            Xp, yp = self.smotetomek.fit_resample(Xp, yp)
        return Xp, yp

    def transform(self, X: np.ndarray) -> np.ndarray:
        Xp = X
        if self.scaler:
            Xp = self.scaler.transform(Xp)
        return Xp
```

```python
# ---------------- Part 6: Base Models ----------------
def build_base_models(
    random_state: int,
    class_weight: Dict[int, float] | None = None,
    scale_pos_weight: float | None = None
):
    # XGBoost (binary uses scale_pos_weight; for multi-class we skip)
    xgb_params = dict(
        n_estimators=300,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        random_state=random_state,
        eval_metric="logloss",
        tree_method="hist"
    )
    if scale_pos_weight is not None:
        xgb_params["scale_pos_weight"] = float(scale_pos_weight)
    xgb = XGBClassifier(**xgb_params)

    # LightGBM (supports class_weight mapping natively)
    lgbm = LGBMClassifier(
        n_estimators=300,
        max_depth=-1,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        random_state=random_state,
        class_weight=class_weight
    )
    return xgb, lgbm
```

```python
# ---------------- Part 7: Ensemble & Calibration ----------------
class EnhancedEnsemble:
    def __init__(self, models: List[Any], method: str = "soft", calibrate: bool = True):
        self.models = models
        self.method = method
        self.calibrate = calibrate
        self.calibrators: List[Any] = []
        self.classes_: np.ndarray | None = None

    def fit(self, X: np.ndarray, y: np.ndarray) -> "EnhancedEnsemble":
        self.classes_ = np.unique(y)
        self.calibrators.clear()
        for mdl in self.models:
            mdl.fit(X, y)
            if self.calibrate:
                calib = CalibratedClassifierCV(mdl, method="isotonic", cv=3)
                calib.fit(X, y)
                self.calibrators.append(calib)
            else:
                self.calibrators.append(mdl)
        return self

    def predict_proba(self, X: np.ndarray) -> np.ndarray:
        prob_list = [c.predict_proba(X) for c in self.calibrators]
        P = np.mean(prob_list, axis=0)  # soft-vote
        return P

    def predict(self, X: np.ndarray) -> np.ndarray:
        P = self.predict_proba(X)
        return self.classes_[np.argmax(P, axis=1)]
```

```python
# ---------------- Part 8: Eval, Persistence & Orchestration ----------------
class MetadataManager:
    def __init__(self, author: str, out_dir: str):
        self.author = author
        self.out_dir = out_dir
        os.makedirs(out_dir, exist_ok=True)

    def save(self, model: Any, cfg: Config, metrics: Dict[str, Any], feature_names: List[str]):
        timestamp = datetime.utcnow().strftime("%Y%m%dT%H%M%SZ")
        base = os.path.join(self.out_dir, f"eeg_model_{timestamp}")
        joblib.dump(model, base + ".joblib")
        with open(base + ".json", "w") as f:
            json.dump({
                "timestamp": timestamp,
                "author": self.author,
                "config": asdict(cfg),
                "metrics": metrics,
                "feature_names": feature_names
            }, f, indent=2)
        return base

def evaluate(y_true: np.ndarray, proba: np.ndarray) -> Dict[str, Any]:
    y_pred = np.argmax(proba, axis=1)
    metrics = {
        "accuracy": float(accuracy_score(y_true, y_pred)),
        "f1_weighted": float(f1_score(y_true, y_pred, average="weighted")),
    }
    # ROC-AUC multi/binary
    try:
        metrics["roc_auc_ovr"] = float(roc_auc_score(y_true, proba, multi_class="ovr"))
    except Exception:
        pass
    return metrics

def plot_confusion(y_true: np.ndarray, y_pred: np.ndarray, save_path: str | None = None) -> None:
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(4,4))
    ax.imshow(cm, interpolation="nearest")
    ax.set_title("Confusion Matrix")
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")
    for (i, j), v in np.ndenumerate(cm):
        ax.text(j, i, str(v), ha="center", va="center")
    fig.tight_layout()
    if save_path: plt.savefig(save_path)
    plt.close(fig)

def run(cfg: Config) -> Dict[str, Any]:
    # Load & validate
    df = load_data(cfg)
    validate(df, cfg.target_col)

    # Feature engineering
    fe = FeatureEngineer()
    df_fe = fe.transform(df, cfg.target_col)

    # Split & weights
    X_train, X_test, y_train, y_test, class_weight, classes, spw = split_and_weights(
        df_fe, cfg.target_col, cfg.test_size, cfg.random_state
    )

    # Preprocess
    pp = Preprocessor(scale=cfg.scale, smote=cfg.smote, random_state=cfg.random_state)
    X_train_p, y_train_p = pp.fit_transform(X_train, y_train)
    X_test_p = pp.transform(X_test)

    # Models & ensemble
    xgb, lgbm = build_base_models(cfg.random_state, class_weight, spw)
    ens = EnhancedEnsemble([xgb, lgbm], method="soft", calibrate=cfg.calibrate).fit(X_train_p, y_train_p)

    # Evaluate
    proba = ens.predict_proba(X_test_p)
    y_pred = np.argmax(proba, axis=1)
    metrics = evaluate(y_test, proba)

    # Persist artifacts + metadata
    meta = MetadataManager(author=cfg.author, out_dir=cfg.model_dir)
    bundle = {"preprocessor": pp, "ensemble": ens}
    base = meta.save(bundle, cfg, metrics, feature_names=fe.feature_names_)

    # Visuals
    plot_confusion(y_test, y_pred, save_path=base + "_cm.png")

    return {"metrics": metrics, "artifact_base": base}

if __name__ == "__main__":
    cfg = Config()
    results = run(cfg)
    print(json.dumps(results, indent=2))
```

```python
# eeg_pipeline_patched.py — dataset-aware, DF-safe
from __future__ import annotations
import os, json
from dataclasses import dataclass, asdict
from typing import Tuple, List, Dict, Any
from datetime import datetime

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    accuracy_score, f1_score, roc_auc_score, confusion_matrix
)
from imblearn.combine import SMOTETomek
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.calibration import CalibratedClassifierCV
import joblib
import matplotlib.pyplot as plt
```

```python
# ---------------- Part 1: Configuration ----------------
@dataclass
class Config:
    data_path: str = "/content/modal_train_data_avg(1).csv"
    target_col: str = "class"
    test_size: float = 0.2
    random_state: int = 42
    scale: bool = True
    smote: bool = True
    model_dir: str = "./artifacts"
    author: str = "user"
    calibrate: bool = True
    calibration_method: str = "sigmoid"  # 'sigmoid' or 'isotonic'
```

```python
# ---------------- Part 2: Data Loading & Validation ----------------
EXPECTED_BANDS = [
    "DELTA","THETA","LOW ALPHA","HIGH ALPHA","LOW BETA","HIGH BETA","LOW GAMMA","HIGH GAMMA"
]
AUX_FEATS = ["GOOD SIGNAL(0-100)","ATTENTION(0-100)","MEDITATION(0-100)"]

def load_data(cfg: Config) -> pd.DataFrame:
    return pd.read_csv(cfg.data_path)

def validate(df: pd.DataFrame, target_col: str) -> None:
    assert target_col in df.columns, f"Missing target column: {target_col}"
    assert set(EXPECTED_BANDS).issubset(df.columns), "Missing one or more EEG band columns."
    # Numeric median impute only
    if df.isna().any().any():
        df.fillna(df.median(numeric_only=True), inplace=True)
```

```python
# ---------------- Part 3: Feature Engineering ----------------
EPS = 1e-6
SIGNAL = "GOOD SIGNAL(0-100)"

class FeatureEngineer:
    def __init__(self):
        self.feature_names_: List[str] | None = None

    def transform(self, df: pd.DataFrame, target_col: str) -> pd.DataFrame:
        X = df.drop(columns=[target_col]).copy()
        y = df[target_col].copy()

        # 1) log1p transform raw band powers (reduce skew)
        band_cols = EXPECTED_BANDS
        for c in band_cols:
            X[c] = np.log1p(np.clip(X[c].values, a_min=0, a_max=None))

        # 2) Alpha/Beta/Gamma averages (in log-space these act like smoothed levels)
        alpha_avg = (X["LOW ALPHA"] + X["HIGH ALPHA"]) / 2.0
        beta_avg  = (X["LOW BETA"]  + X["HIGH BETA"])  / 2.0
        gamma_avg = (X["LOW GAMMA"] + X["HIGH GAMMA"]) / 2.0

        # 3) Ratios (stable after log1p): implemented as differences in log-space
        X["ratio_alpha_beta"]  = (alpha_avg - (beta_avg + 0))          # log(a/b)
        X["ratio_theta_beta"]  = (X["THETA"] - (beta_avg + 0))
        X["ratio_gamma_beta"]  = (gamma_avg - (beta_avg + 0))
        X["ratio_delta_theta"] = (X["DELTA"] - (X["THETA"] + 0))

        # 4) Normalize attention/meditation by signal quality
        X["attn_norm"] = X["ATTENTION(0-100)"]  / (X[SIGNAL] + EPS)
        X["med_norm"]  = X["MEDITATION(0-100)"] / (X[SIGNAL] + EPS)

        # 5) Row-wise stats over transformed band channels
        X["band_mean"] = X[band_cols].mean(axis=1)
        X["band_std"]  = X[band_cols].std(axis=1).fillna(0.0)

        out = pd.concat([X, y], axis=1)
        self.feature_names_ = [c for c in out.columns if c != target_col]
        return out
```

```python
# ---------------- Part 4: Stratified Split & Class Weights ----------------
def split_and_weights(
    df: pd.DataFrame, target_col: str, test_size: float, random_state: int
):
    X_df = df.drop(columns=[target_col])
    y = df[target_col].values

    X_tr, X_te, y_tr, y_te = train_test_split(
        X_df, y, test_size=test_size, random_state=random_state, stratify=y
    )

    classes = np.unique(y_tr)
    weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_tr)
    class_weight = {int(c): float(w) for c, w in zip(classes, weights)}

    scale_pos_weight = None
    if len(classes) == 2:
        w_neg, w_pos = class_weight[int(classes[0])], class_weight[int(classes[1])]
        scale_pos_weight = w_pos / (w_neg + 1e-12)

    return X_tr, X_te, y_tr, y_te, class_weight, classes, scale_pos_weight
```

```python
# ---------------- Part 5: Preprocessing (DF-safe) ----------------
class Preprocessor:
    def __init__(self, scale: bool = True, smote: bool = True, random_state: int = 42):
        self.scale = scale
        self.smote = smote
        self.scaler = StandardScaler() if scale else None
        self.smotetomek = SMOTETomek(random_state=random_state) if smote else None
        self.columns_: List[str] | None = None

    def fit_transform(self, X_df: pd.DataFrame, y: np.ndarray) -> tuple[pd.DataFrame, np.ndarray]:
        self.columns_ = X_df.columns.tolist()
        Xp = X_df.copy()
        if self.scaler:
            Xp[self.columns_] = self.scaler.fit_transform(Xp[self.columns_])
        if self.smotetomek:
            X_arr, y = self.smotetomek.fit_resample(Xp[self.columns_].values, y)
            Xp = pd.DataFrame(X_arr, columns=self.columns_)
        return Xp, y

    def transform(self, X_df: pd.DataFrame) -> pd.DataFrame:
        Xp = X_df.copy()
        if self.scaler:
            Xp[self.columns_] = self.scaler.transform(Xp[self.columns_])
        return Xp
```

```python
# ---------------- Part 6: Base Models ----------------
def build_base_models(
    random_state: int,
    class_weight: Dict[int, float] | None = None,
    scale_pos_weight: float | None = None
):
    # XGBoost: slightly deeper + more trees for capacity
    xgb_params = dict(
        n_estimators=500,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        random_state=random_state,
        eval_metric="mlogloss",
        tree_method="hist"
    )
    if scale_pos_weight is not None:
        xgb_params["scale_pos_weight"] = float(scale_pos_weight)
    xgb = XGBClassifier(**xgb_params)

    # LightGBM: avoid “no positive gain” with more flexible leaves/children
    lgbm = LGBMClassifier(
        n_estimators=400,
        learning_rate=0.05,
        num_leaves=63,
        min_child_samples=20,
        min_split_gain=0.0,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_lambda=1.0,
        random_state=random_state,
        class_weight=class_weight
    )
    return xgb, lgbm
```

```python
# ---------------- Part 7: Ensemble & Calibration ----------------
class EnhancedEnsemble:
    def __init__(self, models: List[Any], method: str = "soft",
                 calibrate: bool = True, calibration_method: str = "sigmoid"):
        self.models = models
        self.method = method
        self.calibrate = calibrate
        self.calibration_method = calibration_method
        self.calibrators: List[Any] = []
        self.classes_: np.ndarray | None = None

    def fit(self, X: pd.DataFrame, y: np.ndarray) -> "EnhancedEnsemble":
        self.classes_ = np.unique(y)
        self.calibrators.clear()
        for mdl in self.models:
            mdl.fit(X, y)
            if self.calibrate:
                calib = CalibratedClassifierCV(mdl, method=self.calibration_method, cv=3)
                calib.fit(X, y)
                self.calibrators.append(calib)
            else:
                self.calibrators.append(mdl)
        return self

    def predict_proba(self, X: pd.DataFrame) -> np.ndarray:
        probs = [c.predict_proba(X) for c in self.calibrators]
        return np.mean(probs, axis=0)  # soft voting

    def predict(self, X: pd.DataFrame) -> np.ndarray:
        P = self.predict_proba(X)
        return self.classes_[np.argmax(P, axis=1)]
```

```python
# ---------------- Part 8: Eval, Persistence & Orchestration ----------------
class MetadataManager:
    def __init__(self, author: str, out_dir: str):
        self.author = author
        self.out_dir = out_dir
        os.makedirs(out_dir, exist_ok=True)

    def save(self, model: Any, cfg: Config, metrics: Dict[str, Any], feature_names: List[str]):
        timestamp = datetime.utcnow().strftime("%Y%m%dT%H%M%SZ")
        base = os.path.join(self.out_dir, f"eeg_model_{timestamp}")
        joblib.dump(model, base + ".joblib")
        with open(base + ".json", "w") as f:
            json.dump({
                "timestamp": timestamp,
                "author": self.author,
                "config": asdict(cfg),
                "metrics": metrics,
                "feature_names": feature_names
            }, f, indent=2)
        return base

def evaluate(y_true: np.ndarray, proba: np.ndarray) -> Dict[str, Any]:
    y_pred = np.argmax(proba, axis=1)
    metrics = {
        "accuracy": float(accuracy_score(y_true, y_pred)),
        "f1_weighted": float(f1_score(y_true, y_pred, average="weighted")),
        "f1_macro": float(f1_score(y_true, y_pred, average="macro")),
    }
    try:
        metrics["roc_auc_ovr"] = float(roc_auc_score(y_true, proba, multi_class="ovr"))
    except Exception:
        pass
    return metrics

def plot_confusion(y_true: np.ndarray, y_pred: np.ndarray, save_path: str | None = None) -> None:
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(4,4))
    ax.imshow(cm, interpolation="nearest")
    ax.set_title("Confusion Matrix")
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")
    for (i, j), v in np.ndenumerate(cm):
        ax.text(j, i, str(v), ha="center", va="center")
    fig.tight_layout()
    if save_path: plt.savefig(save_path)
    plt.close(fig)

def run(cfg: Config) -> Dict[str, Any]:
    # 1) Load & validate
    df = load_data(cfg)
    validate(df, cfg.target_col)

    # 2) Feature engineering
    fe = FeatureEngineer()
    df_fe = fe.transform(df, cfg.target_col)

    # 3) Split & weights (keep DataFrames)
    X_tr, X_te, y_tr, y_te, class_weight, classes, spw = split_and_weights(
        df_fe, cfg.target_col, cfg.test_size, cfg.random_state
    )

    # 4) Preprocess
    pp = Preprocessor(scale=cfg.scale, smote=cfg.smote, random_state=cfg.random_state)
    X_tr_p, y_tr_p = pp.fit_transform(X_tr, y_tr)
    X_te_p = pp.transform(X_te)

    # 5) Models & ensemble
    xgb, lgbm = build_base_models(cfg.random_state, class_weight, spw)
    ens = EnhancedEnsemble(
        [xgb, lgbm], method="soft",
        calibrate=cfg.calibrate, calibration_method=cfg.calibration_method
    ).fit(X_tr_p, y_tr_p)

    # 6) Evaluate
    proba = ens.predict_proba(X_te_p)
    y_pred = np.argmax(proba, axis=1)
    metrics = evaluate(y_te, proba)

    # 7) Persist + visuals
    meta = MetadataManager(author=cfg.author, out_dir=cfg.model_dir)
    base = meta.save({"preprocessor": pp, "ensemble": ens}, cfg, metrics, fe.feature_names_)
    plot_confusion(y_te, y_pred, save_path=base + "_cm.png")

    return {"metrics": metrics, "artifact_base": base}

if __name__ == "__main__":
    cfg = Config()
    results = run(cfg)
    print(json.dumps(results, indent=2))
```

```python
# eeg_pipeline_balanced.py — dataset-aware, class-weighted + targeted oversampling
from __future__ import annotations
import os, json, copy
from dataclasses import dataclass, asdict
from typing import Tuple, List, Dict, Any
from datetime import datetime

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, confusion_matrix, make_scorer
from sklearn.base import clone

from imblearn.over_sampling import RandomOverSampler

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.calibration import CalibratedClassifierCV

import joblib
import matplotlib.pyplot as plt
```

```python
# ---------------- Part 1: Configuration ----------------
@dataclass
class Config:
    data_path: str = "/content/modal_train_data_avg(1).csv"
    target_col: str = "class"
    test_size: float = 0.2
    random_state: int = 42

    # preprocessing
    scale: bool = True

    # targeted oversampling
    enable_targeted_oversample: bool = True
    oversample_classes: List[int] = (3,)   # classes to upsample (tuple/list)
    oversample_factor: float = 1.0         # relative to majority size (≈1.0)

    # modeling
    calibrate: bool = True
    calibration_method: str = "sigmoid"    # 'sigmoid' | 'isotonic'

    # IO
    model_dir: str = "./artifacts"
    author: str = "user"
```

```python
# ---------------- Part 2: Data Loading & Validation ----------------
EXPECTED_BANDS = [
    "DELTA","THETA","LOW ALPHA","HIGH ALPHA","LOW BETA","HIGH BETA","LOW GAMMA","HIGH GAMMA"
]
AUX_FEATS = ["GOOD SIGNAL(0-100)","ATTENTION(0-100)","MEDITATION(0-100)"]

def load_data(cfg: Config) -> pd.DataFrame:
    return pd.read_csv(cfg.data_path)

def validate(df: pd.DataFrame, target_col: str) -> None:
    assert target_col in df.columns, f"Missing target column: {target_col}"
    assert set(EXPECTED_BANDS).issubset(df.columns), "Missing EEG band columns."
    # median-impute numeric NAs (if any)
    if df.isna().any().any():
        df.fillna(df.median(numeric_only=True), inplace=True)
```

```python
# ---------------- Part 3: Feature Engineering ----------------
EPS = 1e-6
SIGNAL = "GOOD SIGNAL(0-100)"

class FeatureEngineer:
    def __init__(self):
        self.feature_names_: List[str] | None = None

    def transform(self, df: pd.DataFrame, target_col: str) -> pd.DataFrame:
        X = df.drop(columns=[target_col]).copy()
        y = df[target_col].copy()

        # (1) log1p transform raw bands to reduce skew
        band_cols = EXPECTED_BANDS
        for c in band_cols:
            X[c] = np.log1p(np.clip(X[c].values, a_min=0, a_max=None))

        # (2) averaged band levels
        alpha_avg = (X["LOW ALPHA"] + X["HIGH ALPHA"]) / 2.0
        beta_avg  = (X["LOW BETA"]  + X["HIGH BETA"])  / 2.0
        gamma_avg = (X["LOW GAMMA"] + X["HIGH GAMMA"]) / 2.0

        # (3) log-ratios (differences in log-space)
        X["ratio_alpha_beta"]  = alpha_avg - beta_avg
        X["ratio_theta_beta"]  = X["THETA"] - beta_avg
        X["ratio_gamma_beta"]  = gamma_avg - beta_avg
        X["ratio_delta_theta"] = X["DELTA"] - X["THETA"]

        # (4) normalize attention/meditation by signal quality
        X["attn_norm"] = X["ATTENTION(0-100)"]  / (X[SIGNAL] + EPS)
        X["med_norm"]  = X["MEDITATION(0-100)"] / (X[SIGNAL] + EPS)

        # (5) row stats across band channels
        X["band_mean"] = X[band_cols].mean(axis=1)
        X["band_std"]  = X[band_cols].std(axis=1).fillna(0.0)

        out = pd.concat([X, y], axis=1)
        self.feature_names_ = [c for c in out.columns if c != target_col]
        return out
```

```python
# ---------------- Part 4: Split & Class Weights ----------------
def split_and_weights(
    df: pd.DataFrame, target_col: str, test_size: float, random_state: int
):
    X_df = df.drop(columns=[target_col])
    y = df[target_col].values

    X_tr, X_te, y_tr, y_te = train_test_split(
        X_df, y, test_size=test_size, random_state=random_state, stratify=y
    )

    classes = np.unique(y_tr)
    weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_tr)
    class_weight = {int(c): float(w) for c, w in zip(classes, weights)}

    # scale_pos_weight for binary XGB only (multi-class uses class_weight through eval set weights)
    scale_pos_weight = None
    if len(classes) == 2:
        w_neg, w_pos = class_weight[int(classes[0])], class_weight[int(classes[1])]
        scale_pos_weight = w_pos / (w_neg + 1e-12)

    return X_tr, X_te, y_tr, y_te, class_weight, classes, scale_pos_weight
```

```python
# ---------------- Part 5: Preprocess & Targeted Oversampling ----------------
class Preprocessor:
    def __init__(self, scale: bool = True, random_state: int = 42):
        self.scale = scale
        self.scaler = StandardScaler() if scale else None
        self.columns_: List[str] | None = None
        self.random_state = random_state

    def fit_transform(self, X_df: pd.DataFrame, y: np.ndarray,
                      enable_os: bool, oversample_classes: List[int],
                      oversample_factor: float) -> tuple[pd.DataFrame, np.ndarray]:
        self.columns_ = X_df.columns.tolist()
        Xp = X_df.copy()
        yp = y.copy()

        # scale (train)
        if self.scaler:
            Xp[self.columns_] = self.scaler.fit_transform(Xp[self.columns_])

        # targeted oversampling (train only)
        if enable_os and len(oversample_classes) > 0:
            # Determine target counts
            cls_counts = pd.Series(yp).value_counts().to_dict()
            majority = max(cls_counts.values())
            sampling: Dict[int, int] = {}
            for c in oversample_classes:
                if c in cls_counts:
                    sampling[c] = int(min(majority, cls_counts[c] + oversample_factor * (majority - cls_counts[c])))
            if sampling:
                ros = RandomOverSampler(sampling_strategy=sampling, random_state=self.random_state)
                X_arr, yp = ros.fit_resample(Xp.values, yp)
                Xp = pd.DataFrame(X_arr, columns=self.columns_)

        return Xp, yp

    def transform(self, X_df: pd.DataFrame) -> pd.DataFrame:
        Xp = X_df.copy()
        if self.scaler:
            Xp[self.columns_] = self.scaler.transform(Xp[self.columns_])
        return Xp
```

```python
# ---------------- Part 6: Base Models ----------------
def build_base_models(
    random_state: int,
    class_weight: Dict[int, float] | None = None,
    scale_pos_weight: float | None = None
):
    # XGBoost
    xgb_params = dict(
        n_estimators=500, max_depth=5, learning_rate=0.05,
        subsample=0.8, colsample_bytree=0.8,
        reg_lambda=1.0, random_state=random_state,
        eval_metric="mlogloss", tree_method="hist"
    )
    if scale_pos_weight is not None:
        xgb_params["scale_pos_weight"] = float(scale_pos_weight)
    xgb = XGBClassifier(**xgb_params)

    # LightGBM (native class_weight)
    lgbm = LGBMClassifier(
        n_estimators=400, learning_rate=0.05,
        num_leaves=63, min_child_samples=20, min_split_gain=0.0,
        subsample=0.8, colsample_bytree=0.8,
        reg_lambda=1.0, random_state=random_state,
        class_weight=class_weight
    )

    # RandomForest (diverse bias; supports class_weight="balanced")
    rf = RandomForestClassifier(
        n_estimators=400, max_depth=None, min_samples_leaf=2,
        class_weight="balanced", n_jobs=-1, random_state=random_state
    )

    return xgb, lgbm, rf
```

```python
# ---------------- Part 7: Ensemble, Calibration & CV-weighted Voting ----------------
class EnhancedEnsemble:
    def __init__(self, models: List[Any], calibrate: bool = True, calibration_method: str = "sigmoid"):
        self.models = models
        self.calibrate = calibrate
        self.calibration_method = calibration_method
        self.calibrators: List[Any] = []
        self.classes_: np.ndarray | None = None
        self.weights_: np.ndarray | None = None  # soft-vote weights

    def _cv_macro_f1_weights(self, X: pd.DataFrame, y: np.ndarray, n_splits: int = 5) -> np.ndarray:
        """Estimate per-model weights from CV macro-F1; normalized to sum=1."""
        scorer = make_scorer(f1_score, average="macro")
        cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
        scores = []
        for mdl in self.models:
            mdl_cv = clone(mdl)
            s = cross_val_score(mdl_cv, X, y, cv=cv, scoring=scorer, n_jobs=None)
            scores.append(max(1e-6, float(np.mean(s))))  # avoid zeros
        w = np.array(scores, dtype=float)
        return w / w.sum()

    def fit(self, X: pd.DataFrame, y: np.ndarray) -> "EnhancedEnsemble":
        self.classes_ = np.unique(y)
        # derive voting weights via CV on macro-F1
        self.weights_ = self._cv_macro_f1_weights(X, y)

        # fit and (optionally) calibrate each model on full train
        self.calibrators.clear()
        for mdl in self.models:
            mdl.fit(X, y)
            if self.calibrate:
                calib = CalibratedClassifierCV(mdl, method=self.calibration_method, cv=3)
                calib.fit(X, y)
                self.calibrators.append(calib)
            else:
                self.calibrators.append(mdl)
        return self

    def predict_proba(self, X: pd.DataFrame) -> np.ndarray:
        probs = [c.predict_proba(X) for c in self.calibrators]
        if self.weights_ is None:
            P = np.mean(probs, axis=0)
        else:
            # weighted soft vote
            P = np.average(probs, axis=0, weights=self.weights_)
        return P

    def predict(self, X: pd.DataFrame) -> np.ndarray:
        P = self.predict_proba(X)
        return self.classes_[np.argmax(P, axis=1)]
```

```python
# ---------------- Part 8: Eval, Persistence & Orchestration ----------------
class MetadataManager:
    def __init__(self, author: str, out_dir: str):
        self.author = author
        self.out_dir = out_dir
        os.makedirs(out_dir, exist_ok=True)

    def save(self, model: Any, cfg: Config, metrics: Dict[str, Any], feature_names: List[str]):
        timestamp = datetime.utcnow().strftime("%Y%m%dT%H%M%SZ")
        base = os.path.join(self.out_dir, f"eeg_model_{timestamp}")
        joblib.dump(model, base + ".joblib")
        with open(base + ".json", "w") as f:
            json.dump({
                "timestamp": timestamp,
                "author": self.author,
                "config": asdict(cfg),
                "metrics": metrics,
                "feature_names": feature_names
            }, f, indent=2)
        return base

def evaluate(y_true: np.ndarray, proba: np.ndarray) -> Dict[str, Any]:
    y_pred = np.argmax(proba, axis=1)
    metrics = {
        "accuracy": float(accuracy_score(y_true, y_pred)),
        "f1_weighted": float(f1_score(y_true, y_pred, average="weighted")),
        "f1_macro": float(f1_score(y_true, y_pred, average="macro")),
    }
    try:
        metrics["roc_auc_ovr"] = float(roc_auc_score(y_true, proba, multi_class="ovr"))
    except Exception:
        pass
    return metrics

def plot_confusion(y_true: np.ndarray, y_pred: np.ndarray, save_path: str | None = None) -> None:
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(4,4))
    ax.imshow(cm, interpolation="nearest")
    ax.set_title("Confusion Matrix")
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")
    for (i, j), v in np.ndenumerate(cm):
        ax.text(j, i, str(v), ha="center", va="center")
    fig.tight_layout()
    if save_path: plt.savefig(save_path)
    plt.close(fig)

def run(cfg: Config) -> Dict[str, Any]:
    # 1) Load & validate
    df = load_data(cfg)
    validate(df, cfg.target_col)

    # 2) Feature engineering
    fe = FeatureEngineer()
    df_fe = fe.transform(df, cfg.target_col)

    # 3) Split & weights
    X_tr, X_te, y_tr, y_te, class_weight, classes, spw = split_and_weights(
        df_fe, cfg.target_col, cfg.test_size, cfg.random_state
    )

    # 4) Preprocess (scale + targeted oversampling)
    pp = Preprocessor(scale=cfg.scale, random_state=cfg.random_state)
    X_tr_p, y_tr_p = pp.fit_transform(
        X_tr, y_tr,
        enable_os=cfg.enable_targeted_oversample,
        oversample_classes=list(cfg.oversample_classes),
        oversample_factor=cfg.oversample_factor
    )
    X_te_p = pp.transform(X_te)

    # 5) Build models
    xgb, lgbm, rf = build_base_models(cfg.random_state, class_weight, spw)

    # 6) Ensemble (CV-weighted soft vote + calibration)
    ens = EnhancedEnsemble(
        [xgb, lgbm, rf],
        calibrate=cfg.calibrate, calibration_method=cfg.calibration_method
    ).fit(X_tr_p, y_tr_p)

    # 7) Evaluate
    proba = ens.predict_proba(X_te_p)
    y_pred = np.argmax(proba, axis=1)
    metrics = evaluate(y_te, proba)

    # 8) Persist
    meta = MetadataManager(author=cfg.author, out_dir=cfg.model_dir)
    base = meta.save({"preprocessor": pp, "ensemble": ens}, cfg, metrics, fe.feature_names_)
    plot_confusion(y_te, y_pred, save_path=base + "_cm.png")

    return {"metrics": metrics, "artifact_base": base}

if __name__ == "__main__":
    cfg = Config()
    results = run(cfg)
    print(json.dumps(results, indent=2))
```

```python
# eeg_pipeline_adasyn_weighted.py
from __future__ import annotations
import os, json
from dataclasses import dataclass, asdict
from typing import Tuple, List, Dict, Any
from datetime import datetime

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler, RobustScaler, QuantileTransformer
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (accuracy_score, f1_score, roc_auc_score,
                             confusion_matrix, make_scorer)
from sklearn.base import clone

from imblearn.over_sampling import ADASYN

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.calibration import CalibratedClassifierCV

import joblib
import matplotlib.pyplot as plt
```

```python
# ---------------- Part 2: Load & Validate ----------------
EXPECTED_BANDS = [
    "DELTA","THETA","LOW ALPHA","HIGH ALPHA","LOW BETA","HIGH BETA","LOW GAMMA","HIGH GAMMA"
]
AUX_FEATS = ["GOOD SIGNAL(0-100)","ATTENTION(0-100)","MEDITATION(0-100)"]

def load_data(cfg: Config) -> pd.DataFrame:
    return pd.read_csv(cfg.data_path)

def validate(df: pd.DataFrame, target_col: str) -> None:
    assert target_col in df.columns, f"Missing target column: {target_col}"
    assert set(EXPECTED_BANDS).issubset(df.columns), "Missing EEG band columns."
    if df.isna().any().any():
        df.fillna(df.median(numeric_only=True), inplace=True)
```

```python
# ---------------- Part 3: Feature Engineering ----------------
EPS = 1e-6
SIGNAL = "GOOD SIGNAL(0-100)"

class FeatureEngineer:
    def __init__(self):
        self.feature_names_: List[str] | None = None

    def transform(self, df: pd.DataFrame, target_col: str) -> pd.DataFrame:
        X = df.drop(columns=[target_col]).copy()
        y = df[target_col].copy()

        # log1p transform raw band powers (reduce skew)
        band_cols = EXPECTED_BANDS
        for c in band_cols:
            X[c] = np.log1p(np.clip(X[c].values, a_min=0, a_max=None))

        # averaged band levels
        alpha_avg = (X["LOW ALPHA"] + X["HIGH ALPHA"]) / 2.0
        beta_avg  = (X["LOW BETA"]  + X["HIGH BETA"])  / 2.0
        gamma_avg = (X["LOW GAMMA"] + X["HIGH GAMMA"]) / 2.0

        # log-ratios (differences)
        X["ratio_alpha_beta"]  = alpha_avg - beta_avg
        X["ratio_theta_beta"]  = X["THETA"] - beta_avg
        X["ratio_gamma_beta"]  = gamma_avg - beta_avg
        X["ratio_delta_theta"] = X["DELTA"] - X["THETA"]

        # normalized aux
        X["attn_norm"] = X["ATTENTION(0-100)"]  / (X[SIGNAL] + EPS)
        X["med_norm"]  = X["MEDITATION(0-100)"] / (X[SIGNAL] + EPS)

        # row stats
        X["band_mean"] = X[band_cols].mean(axis=1)
        X["band_std"]  = X[band_cols].std(axis=1).fillna(0.0)

        out = pd.concat([X, y], axis=1)
        self.feature_names_ = [c for c in out.columns if c != target_col]
        return out
```

```python
# ---------------- Part 4: Split & Class Weights ----------------
def split_and_weights(df: pd.DataFrame, target_col: str,
                      test_size: float, random_state: int):
    X_df = df.drop(columns=[target_col])
    y = df[target_col].values

    X_tr, X_te, y_tr, y_te = train_test_split(
        X_df, y, test_size=test_size, stratify=y, random_state=random_state
    )

    classes = np.unique(y_tr)
    weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_tr)
    class_weight = {int(c): float(w) for c, w in zip(classes, weights)}

    # binary helper (unused for multiclass)
    scale_pos_weight = None
    if len(classes) == 2:
        w_neg, w_pos = class_weight[int(classes[0])], class_weight[int(classes[1])]
        scale_pos_weight = w_pos / (w_neg + 1e-12)

    return X_tr, X_te, y_tr, y_te, class_weight, classes, scale_pos_weight
```

```python
# ---------------- Part 5: Scaling + ADASYN + Row Weights ----------------
class Preprocessor:
    def __init__(self, scaler: str = "standard", random_state: int = 42):
        self.scaler_choice = scaler
        self.columns_: List[str] | None = None
        self.random_state = random_state
        if scaler == "standard":
            self.scaler = StandardScaler()
        elif scaler == "robust":
            self.scaler = RobustScaler()
        elif scaler == "quantile":
            self.scaler = QuantileTransformer(output_distribution="normal", random_state=random_state)
        else:
            raise ValueError("Unknown scaler")

    def _adasyn_strategy_from_factor(
        self, y: np.ndarray, target_classes: Tuple[int, ...], factor: float
    ) -> Dict[int, int]:
        counts = pd.Series(y).value_counts().to_dict()
        if not counts:
            return {}
        majority = max(counts.values())
        strategy: Dict[int, int] = {}
        for c in target_classes:
            if c not in counts:
                continue
            curr = counts[c]
            if curr <= self._adasyn_min_required_neighbors():
                continue
            desired = int(round(curr + factor * (majority - curr)))
            desired = max(curr, min(desired, majority))
            if desired > curr:
                strategy[int(c)] = desired
        return strategy

    @staticmethod
    def _adasyn_min_required_neighbors() -> int:
        return 6  # n_neighbors=5 needs at least 6 minority samples

    def _to_float64_df(self, Xp: pd.DataFrame) -> pd.DataFrame:
        """Ensure all numeric columns are float64 for sklearn/calibration compatibility."""
        Xp = Xp.copy()
        for c in self.columns_:
            Xp[c] = Xp[c].astype(np.float64, copy=False)
        return Xp

    def fit_transform(
        self,
        X_tr: pd.DataFrame,
        y_tr: np.ndarray,
        use_adasyn: bool,
        adasyn_classes: Tuple[int, ...],
        adasyn_neighbors: int,
        adasyn_factor: float,
        class_weight: Dict[int, float],
    ) -> Tuple[pd.DataFrame, np.ndarray, np.ndarray]:
        self.columns_ = X_tr.columns.tolist()
        Xp = X_tr.copy()
        yp = y_tr.astype(np.int32, copy=False)

        # scale
        Xp[self.columns_] = self.scaler.fit_transform(Xp[self.columns_])

        # ADASYN with integer strategy
        if use_adasyn and len(adasyn_classes) > 0:
            strategy = self._adasyn_strategy_from_factor(yp, adasyn_classes, adasyn_factor)
            if strategy:
                ada = ADASYN(sampling_strategy=strategy, n_neighbors=adasyn_neighbors, random_state=self.random_state)
                X_arr, yp = ada.fit_resample(Xp.values, yp)
                Xp = pd.DataFrame(X_arr, columns=self.columns_)

        # <<< DTYPE STANDARDIZATION: features → float64, labels → int32, weights → float64 >>>
        Xp = self._to_float64_df(Xp)
        yp = yp.astype(np.int32, copy=False)
        row_w = np.vectorize(lambda c: float(class_weight[int(c)]))(yp).astype(np.float64, copy=False)

        return Xp, yp, row_w

    def transform(self, X_te: pd.DataFrame) -> pd.DataFrame:
        Xp = X_te.copy()
        Xp[self.columns_] = self.scaler.transform(Xp[self.columns_])
        # keep float64 for inference too
        Xp = self._to_float64_df(Xp)
        return Xp
```

```python
# ---------------- Part 6: Base Models ----------------
def build_base_models(random_state: int,
                      class_weight: Dict[int, float] | None = None,
                      scale_pos_weight: float | None = None):
    # XGBoost
    xgb_params = dict(
        n_estimators=500, max_depth=5, learning_rate=0.05,
        subsample=0.8, colsample_bytree=0.8, reg_lambda=1.0,
        random_state=random_state, eval_metric="mlogloss", tree_method="hist"
    )
    if scale_pos_weight is not None:
        xgb_params["scale_pos_weight"] = float(scale_pos_weight)
    xgb = XGBClassifier(**xgb_params)

    # LightGBM
    lgbm = LGBMClassifier(
        n_estimators=400, learning_rate=0.05,
        num_leaves=63, min_child_samples=20, min_split_gain=0.0,
        subsample=0.8, colsample_bytree=0.8, reg_lambda=1.0,
        random_state=random_state, class_weight=class_weight
    )

    # RandomForest (diverse bias)
    rf = RandomForestClassifier(
        n_estimators=400, max_depth=None, min_samples_leaf=2,
        class_weight="balanced", n_jobs=-1, random_state=random_state
    )
    return xgb, lgbm, rf
```

```python
# ---------------- Part 7: Ensemble (CV macro-F1 weights) + Calibration ----------------
class EnhancedEnsemble:
    def __init__(self, models: List[Any], calibrate: bool = True, calibration_method: str = "sigmoid"):
        self.models = models
        self.calibrate = calibrate
        self.calibration_method = calibration_method
        self.calibrators: List[Any] = []
        self.classes_: np.ndarray | None = None
        self.weights_: np.ndarray | None = None

    def _cv_macro_f1_weights(self, X: pd.DataFrame, y: np.ndarray, n_splits: int = 5) -> np.ndarray:
        scorer = make_scorer(f1_score, average="macro")
        cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
        scores = []
        for mdl in self.models:
            m = clone(mdl)
            s = cross_val_score(m, X, y, cv=cv, scoring=scorer)
            scores.append(max(1e-6, float(np.mean(s))))
        w = np.array(scores, dtype=float)
        return w / w.sum()

    def fit(self, X: pd.DataFrame, y: np.ndarray, sample_weight: np.ndarray) -> "EnhancedEnsemble":
        self.classes_ = np.unique(y)
        self.weights_ = self._cv_macro_f1_weights(X, y)
        self.calibrators.clear()
        # ensure dtypes for all estimators & calibration
        X_fit = X.astype(np.float64, copy=False)
        y_fit = y.astype(np.int32, copy=False)
        sw = sample_weight.astype(np.float64, copy=False)
        for mdl in self.models:
            mdl.fit(X_fit, y_fit, sample_weight=sw)
            if self.calibrate:
                calib = CalibratedClassifierCV(mdl, method=self.calibration_method, cv=3)
                calib.fit(X_fit, y_fit, sample_weight=sw)
                self.calibrators.append(calib)
            else:
                self.calibrators.append(mdl)
        return self

    def predict_proba(self, X: pd.DataFrame) -> np.ndarray:
        # Remove explicit float32 cast for X, rely on Preprocessor output
        probs = [c.predict_proba(X) for c in self.calibrators]
        if self.weights_ is None:
            return np.mean(probs, axis=0)
        return np.average(probs, axis=0, weights=self.weights_)

    def predict(self, X: pd.DataFrame) -> np.ndarray:
        P = self.predict_proba(X)
        return self.classes_[np.argmax(P, axis=1)]
```

```python
# ---------------- Part 8: Threshold Tuning (validation) ----------------
def tune_thresholds(P_val: np.ndarray, y_val: np.ndarray, n_classes: int) -> np.ndarray:
    """
    Simple per-class threshold search to maximize macro-F1 on a validation set.
    Fallback to argmax if no class crosses its threshold.
    """
    best_th = np.full(n_classes, 0.5, dtype=float)
    best_score = -1.0
    # coarse grid; can refine later around winners
    grid = [0.3, 0.4, 0.5, 0.6]
    from itertools import product
    for th in product(grid, repeat=n_classes):
        th = np.array(th, dtype=float)
        # one-vs-rest decision with fallback
        pred_bin = (P_val >= th).astype(int)
        y_hat = np.where(pred_bin.sum(axis=1) == 0, P_val.argmax(axis=1), pred_bin.argmax(axis=1))
        s = f1_score(y_val, y_hat, average="macro")
        if s > best_score:
            best_score, best_th = s, th
    return best_th

def apply_thresholds(P: np.ndarray, th: np.ndarray) -> np.ndarray:
    pred_bin = (P >= th).astype(int)
    return np.where(pred_bin.sum(axis=1) == 0, P.argmax(axis=1), pred_bin.argmax(axis=1))
```

```python
# ---------------- Part 9: Eval, Persistence & Orchestration ----------------
class MetadataManager:
    def __init__(self, author: str, out_dir: str):
        self.author = author
        self.out_dir = out_dir
        os.makedirs(out_dir, exist_ok=True)

    def save(self, model: Any, cfg: Config, metrics: Dict[str, Any],
             feature_names: List[str], thresholds: List[float]):
        timestamp = datetime.utcnow().strftime("%Y%m%dT%H%M%SZ")
        base = os.path.join(self.out_dir, f"eeg_model_{timestamp}")
        joblib.dump(model, base + ".joblib")
        with open(base + ".json", "w") as f:
            json.dump({
                "timestamp": timestamp,
                "author": self.author,
                "config": asdict(cfg),
                "metrics": metrics,
                "feature_names": feature_names,
                "thresholds": list(map(float, thresholds))
            }, f, indent=2)
        return base

def evaluate(y_true: np.ndarray, proba: np.ndarray, th: np.ndarray | None = None) -> Dict[str, Any]:
    if th is None:
        y_pred = np.argmax(proba, axis=1)
    else:
        y_pred = apply_thresholds(proba, th)
    metrics = {
        "accuracy": float(accuracy_score(y_true, y_pred)),
        "f1_weighted": float(f1_score(y_true, y_pred, average="weighted")),
        "f1_macro": float(f1_score(y_true, y_pred, average="macro")),
    }
    try:
        metrics["roc_auc_ovr"] = float(roc_auc_score(y_true, proba, multi_class="ovr"))
    except Exception:
        pass
    return metrics

def plot_confusion(y_true: np.ndarray, y_pred: np.ndarray, save_path: str | None = None) -> None:
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(4,4))
    ax.imshow(cm, interpolation="nearest")
    ax.set_title("Confusion Matrix")
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")
    for (i, j), v in np.ndenumerate(cm):
        ax.text(j, i, str(v), ha="center", va="center")
    fig.tight_layout()
    if save_path: plt.savefig(save_path)
    plt.close(fig)

def run(cfg: Config) -> Dict[str, Any]:
    # 1) Load & validate
    df = load_data(cfg)
    validate(df, cfg.target_col)

    # 2) Features
    fe = FeatureEngineer()
    df_fe = fe.transform(df, cfg.target_col)

    # 3) Split & weights
    X_tr, X_te, y_tr, y_te, class_weight, classes, spw = split_and_weights(
        df_fe, cfg.target_col, cfg.test_size, cfg.random_state
    )

    # 4) Small validation from train
    X_tr_in, X_val, y_tr_in, y_val = train_test_split(
        X_tr, y_tr, test_size=0.15, stratify=y_tr, random_state=cfg.random_state
    )

    # 5) Preprocess (fit on train-in), then transform VAL and TEST
    pp = Preprocessor(scaler=cfg.scaler, random_state=cfg.random_state)
    X_tr_p, y_tr_p, row_w = pp.fit_transform(
        X_tr_in, y_tr_in,
        use_adasyn=cfg.use_adasyn,
        adasyn_classes=cfg.adasyn_classes,
        adasyn_neighbors=cfg.adasyn_neighbors,
        adasyn_factor=cfg.adasyn_factor,
        class_weight=class_weight
    )
    X_val_p = pp.transform(X_val)
    X_te_p  = pp.transform(X_te)
    assert isinstance(X_te_p, pd.DataFrame) and len(X_te_p) == len(X_te), \
        "Test features not preprocessed."

    # 6) Models
    xgb, lgbm, rf = build_base_models(cfg.random_state, class_weight, spw)

    # 7) Ensemble (fit with row weights)
    ens = EnhancedEnsemble(
        [xgb, lgbm, rf],
        calibrate=cfg.calibrate,
        calibration_method=cfg.calibration_method
    ).fit(X_tr_p, y_tr_p, sample_weight=row_w)

    # 8) Tune thresholds on validation
    P_val = ens.predict_proba(X_val_p)
    th = tune_thresholds(P_val, y_val, n_classes=len(np.unique(y_tr)))

    # 9) Evaluate on TEST
    P_te = ens.predict_proba(X_te_p)
    y_pred = apply_thresholds(P_te, th)
    metrics = evaluate(y_te, P_te, th=th)

    # 10) Persist
    meta = MetadataManager(author=cfg.author, out_dir=cfg.model_dir)
    base = meta.save({"preprocessor": pp, "ensemble": ens}, cfg, metrics, fe.feature_names_, th)
    plot_confusion(y_te, y_pred, save_path=base + "_cm.png")
    return {"metrics": metrics, "artifact_base": base}

if __name__ == "__main__":
    cfg = Config()
    results = run(cfg)
    print(json.dumps(results, indent=2))
```
